```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 读取 Excel 文件
df = pd.read_csv("D:\\FDSM\\Coding\\Computer Program and Data Visualization\\Pre\\Global_Space_Exploration_Dataset.csv")

# 构建合作矩阵（国家之间的合作频率）
# 假设每个任务的“Collaborating Countries”列是一个用逗号分隔的国家列表
df['Collaborating Countries'] = df['Collaborating Countries'].str.split(',')

# 为每个任务记录所有合作国家
countries = set([country for sublist in df['Collaborating Countries'] for country in sublist])
country_dict = {country: [] for country in countries}

for index, row in df.iterrows():
    countries_in_mission = set(row['Collaborating Countries'])
    for country in country_dict:
        country_dict[country].append(1 if country in countries_in_mission else 0)

# 将合作矩阵转换为 DataFrame
collab_matrix = pd.DataFrame(country_dict)

# 绘制热力图
plt.figure(figsize=(12, 8))
sns.heatmap(collab_matrix.corr(), annot=False, cmap='YlGnBu', linewidths=0.5, cbar=True)
plt.title('Correlation Heatmap of Countries Collaboration', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

# 创建技术与国家的交叉表
tech_country = pd.crosstab(df['Technology Used'], df['Country'])

# 将交叉表转换为适合柱状图的格式
tech_country_melted = tech_country.reset_index().melt(
    id_vars='Technology Used', 
    var_name='Country', 
    value_name='Count'
)

# 设置图表尺寸和样式
plt.figure(figsize=(14, 8))
plt.style.use('seaborn-v0_8-whitegrid')

# 创建分组柱状图
ax = sns.barplot(
    x='Technology Used', 
    y='Count', 
    hue='Country', 
    data=tech_country_melted,
    palette='rocket'
)

# 美化图表
ax.set_title('不同国家使用各类航天技术的频率', fontsize=18, fontweight='bold', pad=20)
ax.set_xlabel('技术类型', fontsize=14, fontweight='bold')
ax.set_ylabel('使用次数', fontsize=14, fontweight='bold')

# 旋转x轴标签以避免重叠
plt.xticks(rotation=45, ha='right')

# 优化图例
plt.legend(title='国家', bbox_to_anchor=(1.05, 1), loc='upper left')

# 添加数据标签（如果数据点不太多的情况下）
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3)

# 调整布局并保存
plt.tight_layout()
plt.savefig('tech_country_barchart.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

# 计算每个国家的任务平均持续天数
country_duration = df.groupby('Country')['Duration (in Days)'].mean().reset_index()

# 按平均持续时间降序排序
country_duration = country_duration.sort_values('Duration (in Days)', ascending=False)

# 设置图表尺寸和样式
plt.figure(figsize=(12, 7))
plt.style.use('seaborn-v0_8-whitegrid')

# 创建柱状图
bars = plt.bar(
    country_duration['Country'], 
    country_duration['Duration (in Days)'],
    color=sns.color_palette("rocket", len(country_duration))
)

# 美化图表
plt.title('各国航天任务平均持续时间对比', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('国家', fontsize=14, fontweight='bold')
plt.ylabel('平均持续时间 (天)', fontsize=14, fontweight='bold')

# 添加数据标签
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2., 
        height + 5,
        f'{height:.1f}',
        ha='center', 
        va='bottom',
        fontsize=12,
        fontweight='bold'
    )

# 添加网格线使数值更易读
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 调整轴范围确保标签可见
plt.ylim(0, country_duration['Duration (in Days)'].max() * 1.15)

# 添加平均线
avg_duration = df['Duration (in Days)'].mean()
plt.axhline(y=avg_duration, color='red', linestyle='--', alpha=0.7)
plt.text(
    len(country_duration) - 1, 
    avg_duration + 5, 
    f'总体平均: {avg_duration:.1f}天',
    color='red',
    fontsize=10,
    fontweight='bold'
)

# 调整布局并保存
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 将协作国家列拆分为列表
df['Collab_List'] = df['Collaborating Countries'].str.split(',')

# 计算每个国家参与合作的次数
collaboration_counts = {}
for idx, row in df.iterrows():
    country = row['Country']
    # 确保协作列表存在且不为空
    if isinstance(row['Collab_List'], list) and row['Collab_List'][0] != '':
        if country not in collaboration_counts:
            collaboration_counts[country] = len(row['Collab_List'])
        else:
            collaboration_counts[country] += len(row['Collab_List'])

# 转换为DataFrame并排序
collab_df = pd.DataFrame(list(collaboration_counts.items()), columns=['Country', 'Collaboration Count'])
collab_df = collab_df.sort_values('Collaboration Count', ascending=False)

plt.figure(figsize=(12, 6))
plt.bar(collab_df['Country'], collab_df['Collaboration Count'], color='purple')
plt.title('各国国际合作参与度', fontsize=14)
plt.xlabel('国家')
plt.ylabel('合作次数')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('collaboration_count.png')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 假设你已经加载了数据集，命名为 df
# 将数据按国家和年份聚合，计算预算总和和成功率平均值
df_agg = df.groupby(['Country', 'Year']).agg({
    'Budget (in Billion $)': 'sum',
    'Success Rate (%)': 'mean'
}).reset_index()

# 将数据透视化，绘制热力图
pivot_budget = df_agg.pivot('Country', 'Year', 'Budget (in Billion $)')
pivot_success_rate = df_agg.pivot('Country', 'Year', 'Success Rate (%)')

# 绘制热力图
fig, ax = plt.subplots(1, 2, figsize=(18, 6))

sns.heatmap(pivot_budget, annot=False, cmap='YlGnBu', ax=ax[0])
ax[0].set_title('Budget by Country and Year')

sns.heatmap(pivot_success_rate, annot=False, cmap='YlOrRd', ax=ax[1])
ax[1].set_title('Success Rate by Country and Year')

plt.tight_layout()
plt.show()
```

```python
# 绘制环境影响与任务成功率的关系（小提琴图）
plt.figure(figsize=(12, 6))
sns.violinplot(x='Environmental Impact', y='Success Rate (%)', hue='Mission Type', data=df, split=True, palette='muted')

plt.title('Success Rate Distribution by Environmental Impact and Mission Type')
plt.xlabel('Environmental Impact')
plt.ylabel('Success Rate (%)')
plt.legend(title='Mission Type')
plt.show()
```

```python
# 按年份和任务类型聚合任务数量
df_agg_year_type = df.groupby(['Year', 'Mission Type']).size().unstack(fill_value=0)

# 绘制堆叠面积图
df_agg_year_type.plot(kind='area', stacked=True, figsize=(12, 6), colormap='Set2')

plt.title('Trend of Mission Types over Years')
plt.xlabel('Year')
plt.ylabel('Number of Missions')
plt.show()
```

```python
# 按年份和任务类型聚合预算和成功率
df_agg_year = df.groupby(['Year', 'Mission Type']).agg({
    'Budget (in Billion $)': 'sum',
    'Success Rate (%)': 'mean'
}).reset_index()

# 绘制堆积线图
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Budget (in Billion $)', hue='Mission Type', data=df_agg_year, marker='o')
plt.title('Total Budget by Mission Type Over Time')
plt.xlabel('Year')
plt.ylabel('Total Budget (in Billion $)')
plt.legend(title='Mission Type')
plt.show()

# 绘制任务类型的成功率变化趋势
plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Success Rate (%)', hue='Mission Type', data=df_agg_year, marker='o')
plt.title('Success Rate Over Time by Mission Type')
plt.xlabel('Year')
plt.ylabel('Success Rate (%)')
plt.legend(title='Mission Type')
plt.show()
```

```python
import matplotlib.pyplot as plt

# 按卫星种类和任务类型聚合预算
df_agg_budget = df.groupby(['Satellite Type', 'Mission Type'])['Budget (in Billion $)'].sum().unstack(fill_value=0)

# 绘制堆叠柱状图
ax = df_agg_budget.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='Set2')

# 添加标题和标签
plt.title('Total Budget by Satellite Type and Mission Type')
plt.xlabel('Satellite Type')
plt.ylabel('Total Budget (in Billion $)')
plt.xticks(rotation=45)
plt.legend(title='Mission Type')

# 显示每个堆叠部分的数值
for p in ax.patches:
    height = p.get_height()  # 每个矩形条的高度（即预算数值）
    width = p.get_width()    # 每个矩形条的宽度
    x = p.get_x() + width / 2  # x坐标，设置为条形图的中心位置
    y = p.get_y() + height / 2  # y坐标，设置为条形图的中心位置
    ax.text(x, y, f'{height:.2f}', ha='center', va='center', color='white', fontweight='bold')

# 显示图表
plt.show()
```
